```python
output_csv_path = './papers_data_from_symantic_scholar_Arabic world_2.csv'
```

```python
with open(output_csv_path, mode='w', newline='', encoding='utf-8') as output_file:
    fieldnames = ['Paper Id' , 'title', 'authors', 'citationCount', 'references', 'year', 'isOpenAccess', 'externalIds', 'publicationDate', 'embedding']
    writer = csv.DictWriter(output_file, fieldnames=fieldnames)
    writer.writeheader()

    for i in range (0,len(papers_dois),500):  
        url = 'https://api.semanticscholar.org/graph/v1/paper/batch'
        query_params = {"fields": "title,authors,citationCount,references,url,fieldsOfStudy,year,isOpenAccess,externalIds,paperId,publicationDate,embedding"}
        api_key = "9zJhKmoKKZ5OyKBkPj6rK6zWa1oxTizegHHhNT6g"
        headers = {"x-api-key": api_key}
        papers_details = requests.post(
            url,
            params= query_params,
            json={"ids": papers_dois[i:i+500]},
            headers=headers
        )

        try:
            papers_details = papers_details.json()
        except ValueError as e:
            print(f"Error parsing JSON at batch {i}: {e}")
            continue

        if not isinstance(papers_details, list):
            print(f"Unexpected response format at batch {i}: {papers_details}")
            continue

        for paper in papers_details:
            if paper:
                data_to_save = {
                    'Paper Id': paper["paperId"],
                    'title': paper["title"],
                    'authors': paper["authors"],
                    'citationCount': paper["citationCount"],
                    'references': paper["references"],
                    'year': paper["year"],
                    'isOpenAccess': paper["isOpenAccess"],
                    'externalIds': paper["externalIds"],
                    'publicationDate': paper["publicationDate"],
                    'embedding': paper["embedding"]
                }
                writer.writerow(data_to_save)

        print(f"prossesed {i + 500} papers")
```
